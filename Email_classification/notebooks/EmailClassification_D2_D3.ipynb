## Stage 3: Dataset Registry & Normalized Loading

### 3.1 Dataset Registry & Provenance

The D2 (Enron-Spam) and D3 (SpamAssassin) datasets were prepared by `src/prepare_d2_d3.py`

Normalized CSV files at `data/d2_d3_normalized/enron_spam.csv` and `data/d2_d3_normalized/spamassassin.csv`.

````python
# ============================================================
# STAGE 3: DATASET REGISTRY & NORMALIZED LOADER
# ============================================================
import hashlib

# ---------------------------------------------------------------
# DATASET REGISTRY
# ---------------------------------------------------------------
DATASETS = {
    "enron_spam": {
        "path":             DATA_DIR / "d2_d3_normalized" / "enron_spam.csv",
        "task":             "binary_spam",
        "source":           "Enron-Spam corpus (SetFit/enron_spam on HuggingFace)",
        "label_map_original": {"ham": "legitimate", "spam": "spam"},
        "label_map": {"legitimate": "Non-spam (ham) email", "spam": "Unsolicited bulk email"},
        "license":          "Research / education use; original Enron corpus public.",
        "known_limitations": "Enron era (early 2000s); corporate email style; may not reflect modern spam.",
    },
    "spamassassin": {
        "path":             DATA_DIR / "d2_d3_normalized" / "spamassassin.csv",
        "task":             "binary_spam",
        "source":           "Apache SpamAssassin public corpus (bvk/SpamAssassin-spam on HuggingFace)",
        "label_map_original": {"0": "legitimate", "1": "spam"},
        "label_map": {"legitimate": "Non-spam (ham) email", "spam": "Unsolicited bulk email"},
        "license":          "Education/research per SpamAssassin project.",
        "known_limitations": "Older corpus; some emails truncated; mixed languages.",
    },
}

# ---------------------------------------------------------------
# NORMALISED LOADER
# ---------------------------------------------------------------
def load_dataset(dataset_id, config):
    """
    Load a normalized CSV file, validate its schema, normalise text fields.

    Parameters
    ----------
    dataset_id : str       e.g. "enron_spam"
    config     : dict       registry entry: path, expected_columns

    Returns
    -------
    pd.DataFrame  with columns email_id, subject, body, label,
                  text, text_length, dataset_id.

    Raises
    ------
    FileNotFoundError  CSV not found.
    ValueError          Required columns absent.
    """
    path = config["path"]
    if not path.exists():
        raise FileNotFoundError(
            f"[{dataset_id}] Missing file: {path.resolve()}")

    df_raw = pd.read_csv(path)
    print(f"\n{'='*70}")
    print(f"[{dataset_id}] Loaded {path.name}  rows={len(df_raw)}  "
          f"columns={list(df_raw.columns)}")

    # Hard-fail schema validation
    missing = REQUIRED_COLUMNS.difference(df_raw.columns)
    if missing:
        raise ValueError(
            f"[{dataset_id}] CRITICAL: missing columns {sorted(missing)}! "
            f"Schema requires {sorted(REQUIRED_COLUMNS)}.")

    df = df_raw.copy()

    # Normalise text fields
    df["subject"] = df["subject"].fillna("").astype(str)
    df["body"]    = df["body"].fillna("").astype(str)
    df["label"]   = df["label"].astype(str).str.strip().str.lower()
    df["dataset_id"] = dataset_id

    # Build unified 'text' field, following manual section 11.2
    df["text"] = (
        "subject: " + df["subject"].str.strip()
        + "\nbody: " + df["body"].str.strip()
    )
    df["text_length"] = df["text"].str.len()

    # Verify binary label space is harmonized
    unique_labels = sorted(df["label"].unique())
    expected_binary = ["legitimate", "spam"]
    if unique_labels != expected_binary:
        print(f"  WARNING: Labels = {unique_labels}, expected {expected_binary}")

    print(f"  classes: {unique_labels}  |  "
          f"empty_text: {(df['text'].str.strip()=='').sum()}")
    print(f"  duplicate text rows: {df['text'].duplicated().sum()}  |  "
          f"median len: {df['text_length'].median():.0f}")
    print(f"{'='*70}\n")
    return df


# ---- Load all datasets ----
datasets = {}
for ds_id, cfg in DATASETS.items():
    datasets[ds_id] = load_dataset(ds_id, cfg)

df_d2 = datasets["enron_spam"]
df_d3 = datasets["spamassassin"]

print("[SUMMARY] D2 (Enron-Spam) head:")
display(df_d2.head(5))
print("\n[SUMMARY] D3 (SpamAssassin) head:")
display(df_d3.head(5))
````

## Stage 4: Data Audit & Quality Assessment

### 4.1 Reproducible Hashing & Structured Audit

Each dataset is audited **before** modelling. The audit table captures:

- Row count, class cardinality, and label balance
- Empty / whitespace-only text entries
- Exact duplicate texts (including campaign repeats)
- Text-length descriptive statistics
- SHA-256 hash of the source CSV for immutability tracking

### 4.2 Duplicate Policy (Required Audit Decision)

> *"Determine whether they are repeated messages, quoted threads, spam campaigns, or accidental duplicate rows. Do not silently delete duplicate emails."*
> -- Manual section 11.3

| Pattern | Likely Cause | Action |
|---------|------------|--------|
| Duplicate rows, same label | Spam campaign repetition | **Retain**, document reasons |
| Accidental duplicate rows | Data-entry error | Remove & count |
| Thread / quoted-history duplication | Leakage across train/test | Remove (when group IDs exist) |

For D2 & D3: Campaign-level duplicates are expected in spam corpora.
Decision: **RETAIN** duplicates; fully documented (disclosure item for report).

````python
# ============================================================
# STAGE 4: DATASET AUDIT HASH + TABLE + DUP ANALYSIS
# ============================================================

def sha256_file(path, chunk_size=1024 * 1024):
    """
    Compute SHA-256 hex digest of a file via chunked reading (safe for large files).
    """
    digest = hashlib.sha256()
    with open(path, "rb") as fh:
        for chunk in iter(lambda: fh.read(chunk_size), b""):
            digest.update(chunk)
    return digest.hexdigest()


def audit_dataset(dataset_id, df, source_path):
    """
    Produce a structured audit record for a single dataset.

    Returns
    -------
    dict  Keys: dataset_id, rows, classes, empty_text, exact_duplicate_text,
                 unique_texts, median_text_length, file_sha256
    """
    empty_cnt = int((df["text"].str.strip() == "").sum())
    dup_cnt   = int(df["text"].duplicated(keep="first").sum())
    return {
        "dataset_id":          dataset_id,
        "rows":                len(df),
        "classes":             df["label"].nunique(),
        "empty_text":          empty_cnt,
        "exact_duplicate_text": dup_cnt,
        "unique_texts":        df["text"].nunique(),
        "median_text_length":  float(df["text_length"].median()),
        "file_sha256":         sha256_file(source_path),
    }


# ---- Execute audits ----
audit_records = []
for ds_id, df in datasets.items():
    cfg = DATASETS[ds_id]
    audit_records.append(audit_dataset(ds_id, df, cfg["path"]))

audit_df = pd.DataFrame(audit_records)
print("\n[AUDIT] Dataset quality summary (pre-cleaning):")
display(audit_df)

# ---- Per-dataset label distribution ----
for ds_id, df in datasets.items():
    print(f"\n[AUDIT] Label distribution -> {ds_id}:")
    cnts = df["label"].value_counts(dropna=False)
    pct  = (cnts / len(df) * 100).round(1)
    dist = pd.DataFrame({"count": cnts, "%": pct})
    dist.index.name = "label"
    display(dist)

# ---- Deep duplicate investigation ----
for ds_id, df in datasets.items():
    print(f"\n[AUDIT] Duplicate text analysis -> {ds_id}:")
    df_t = df.copy()
    df_t["is_dup"] = df_t["text"].duplicated(keep="first")
    dup_by_class = df_t[df_t["is_dup"]]["label"].value_counts()

    print(f"  Duplicate rows (after first): {df_t['is_dup'].sum()}")
    print(f"  Unique texts:                {df_t['text'].nunique()}")
    print("\n  Duplicates by class:")
    for label, cnt in sorted(dup_by_class.items()):
        print(f"    {label}: {cnt}")

    # Label-conflict check
    n_conflicts = (df_t.groupby("text")["label"].nunique() > 1).sum()
    print(f"\n  Texts with label conflict: {n_conflicts}")
    if n_conflicts == 0:
        print(
            "  -> No conflicts - consistent with campaign/template repetition.\n"
            "  -> Decision: KEEP, document fully (manual section 11.3).")
    else:
        print("  -> WARNING: LABEL CONFLICT - removal/reconciliation required.")

# ---- Persist audit CSV ----
audit_path = ARTIFACT_DIR / "dataset_audit_d2_d3.csv"
audit_df.to_csv(audit_path, index=False)
print(f"\n[OK] Audit saved -> {audit_path.resolve()}")
````

## Stage 5: Leakage-Safe Train / Test Split

### 5.1 Stratified, Locked Split Protocol

The train/test split is created **once** and locked. All subsequent CV and
hyper-parameter exploration uses only the **training** partition. The test set
is evaluated exactly once after the final model is chosen.

- **Seed:** `RANDOM_STATE = 42`
- **Stratification:** on `label` for proportional class balance
- **Ratio:** 80% train / 20% test
- **Manifest:** persisted to `artifacts/split_manifest_d2_d3.json` for reproducibility

Neither D2 nor D3 supplies `thread_id` or `sender_group`; standard stratified
split is used. If available, grouped split would be preferred to prevent
campaign leakage.

````python
# ============================================================
# STAGE 5: LOCKED, STRATIFIED TRAIN/TEST SPLIT
# ============================================================
from sklearn.model_selection import train_test_split

def make_split(df, test_size=TEST_SIZE, random_state=RANDOM_STATE):
    """
    Return train_df, test_df with a fixed seed and label-stratification.
    Indices are reset so slices are contiguous.
    """
    train, test = train_test_split(
        df,
        test_size=test_size,
        random_state=random_state,
        stratify=df["label"],
    )
    return train.reset_index(drop=True), test.reset_index(drop=True)


splits = {}
for ds_id, df in datasets.items():
    tr, te = make_split(df)
    splits[ds_id] = {"train": tr, "test": te}
    print(f"\n[SPLIT] {ds_id:22s}  train={len(tr):6d}  test={len(te):6d}")
    print("  Train labels: ", end="")
    print(", ".join(f"{k}:{v}" for k, v in sorted(tr["label"].value_counts().items())))
    print("  Test  labels: ", end="")
    print(", ".join(f"{k}:{v}" for k, v in sorted(te["label"].value_counts().items())))

# ---- Persist manifest (email_id list) ----
manifest = {
    ds_id: {
        "train_ids": sp["train"]["email_id"].tolist(),
        "test_ids":  sp["test"]["email_id"].tolist(),
    }
    for ds_id, sp in splits.items()
}

manifest_path = ARTIFACT_DIR / "split_manifest_d2_d3.json"
with open(manifest_path, "w", encoding="utf-8") as fh:
    json.dump(manifest, fh, indent=2, ensure_ascii=False)
print(f"\n[SER] Split manifest -> {manifest_path.resolve()}")
````

## Stage 6: TF-IDF Pipeline Registry

### 6.1 Five classifiers over a common TF-IDF front-end

`TfidfVectorizer` is placed inside a `Pipeline` so that vocabulary and IDF
statistics are fitted **only on training folds** during cross-validation
-- which prevents both vocabulary leakage and IDF leakage.

### 6.2 Required five classifiers

| Classifier | Role |
|----------|------|
| `DummyClassifier` (most_frequent) | Performance floor |
| `MultinomialNB` (alpha=1.0) | Fast sparse-text baseline |
| `ComplementNB` (alpha=1.0) | Better imbalanced-text baseline |
| `LogisticRegression` (C=1.0, class_weight='balanced') | Linear discriminative boundary |
| `LinearSVC` (class_weight='balanced') | Maximum-margin for sparse high-dim text |

````python
# ============================================================
# STAGE 6: PIPELINE REGISTRY (TF-IDF + classifiers)
# ============================================================
from sklearn.dummy import DummyClassifier
from sklearn.naive_bayes import MultinomialNB, ComplementNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC

def make_pipeline(classifier):
    """
    Wrap a TF-IDF vectoriser + classifier into a single Pipeline.
    During CV, the vectorizer is fitted only on training folds -- no leakage.
    """
    return Pipeline([
        ("tfidf", TfidfVectorizer(**TFIDF_CONFIG)),
        ("classifier", classifier),
    ])


MODELS = {
    "dummy_majority": make_pipeline(
        DummyClassifier(strategy="most_frequent")
    ),
    "multinomial_nb": make_pipeline(
        MultinomialNB(alpha=1.0)
    ),
    "complement_nb": make_pipeline(
        ComplementNB(alpha=1.0)
    ),
    "logistic_regression": make_pipeline(
        LogisticRegression(
            max_iter=2500,
            C=1.0,
            class_weight="balanced",
            random_state=RANDOM_STATE,
        )
    ),
    "linear_svc": make_pipeline(
        LinearSVC(
            class_weight="balanced",
            random_state=RANDOM_STATE,
        )
    ),
}

print(f"[PIPELINE] {len(MODELS)} classifiers registered:")
for name, pipe in MODELS.items():
    print(f"  . {name:30s} (estimator steps: {list(pipe.named_steps.keys())})")
````

## Stage 7: Training-Only Cross-Validation Benchmark

### 7.1 Five-Fold Stratified CV Evaluation

We run 5-fold stratified cross-validation on the training partition **only**.
No test data is touched. Metrics tracked: `accuracy`, `macro_f1`, `weighted_f1`.

This is performed **independently for each dataset** (D2 and D3).

````python
# ============================================================
# STAGE 7: TRAINING-ONLY CROSS-VALIDATION BENCHMARK (D2 & D3)
# ============================================================
from sklearn.model_selection import StratifiedKFold, cross_validate

SCORING = {
    "accuracy": "accuracy",
    "macro_f1": "f1_macro",
    "weighted_f1": "f1_weighted",
}

cv = StratifiedKFold(
    n_splits=CV_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE,
)

cv_rows = []

for ds_id, part in splits.items():
    train_df = part["train"]
    X_train = train_df["text"]
    y_train = train_df["label"]

    print(f"\n[CV] Running 5-fold CV on {ds_id} (train={len(train_df)}) ...")

    for model_name, pipeline in MODELS.items():
        scores = cross_validate(
            pipeline,
            X_train,
            y_train,
            cv=cv,
            scoring=SCORING,
            n_jobs=N_JOBS,
            error_score="raise",
        )
        cv_rows.append({
            "dataset_id": ds_id,
            "model": model_name,
            "accuracy_mean": float(scores["test_accuracy"].mean()),
            "accuracy_sd": float(scores["test_accuracy"].std()),
            "macro_f1_mean": float(scores["test_macro_f1"].mean()),
            "macro_f1_sd": float(scores["test_macro_f1"].std()),
            "weighted_f1_mean": float(scores["test_weighted_f1"].mean()),
            "weighted_f1_sd": float(scores["test_weighted_f1"].std()),
        })

cv_results = pd.DataFrame(cv_rows)
cv_results = cv_results.sort_values(["dataset_id", "macro_f1_mean"], ascending=[True, False]).reset_index(drop=True)

print("\n[CV] Training-only Cross-Validation Benchmark (D2 & D3):")
display(cv_results)

cv_csv_path = ARTIFACT_DIR / "cv_results_d2_d3.csv"
cv_results.to_csv(cv_csv_path, index=False)
print(f"[OK] CV results saved -> {cv_csv_path.resolve()}")
````

### 7.3 Overfitting Check (Train vs. CV Macro F1)

We estimate the train-vs-CV macro F1 gap **inside the training partition only**. A gap larger than `OVERFIT_GAP = 0.05` flags a model as potentially overfit; we then prefer the simpler / more regularised alternative among practically tied models.

````python
# ============================================================
# STAGE 7.3: OVERFITTING CHECK (TRAIN vs CV MACRO F1 GAP)
# ============================================================
OVERFIT_GAP = 0.05  # flag if train-macro-F1 - cv-macro-F1 > OVERFIT_GAP

def train_macro_f1(pipeline, X, y):
    """Fit on the full training set, score macro F1 on the same data.
    This is a deliberately optimistic upper bound used to spot
    overfitting, not for model selection."""
    pipe = clone(pipeline)
    pipe.fit(X, y)
    return f1_score(y, pipe.predict(X), average="macro")

for ds_id in splits.keys():
    train_df = splits[ds_id]["train"]
    X_train = train_df["text"]
    y_train = train_df["label"]
    ds_cv = cv_results[cv_results["dataset_id"] == ds_id].reset_index(drop=True)

    overfit_rows = []
    for _, row in ds_cv.iterrows():
        name = row["model"]
        cv_f1 = float(row["macro_f1_mean"])
        train_f1 = train_macro_f1(MODELS[name], X_train, y_train)
        overfit_rows.append({
            "model": name,
            "train_macro_f1": train_f1,
            "cv_macro_f1":    cv_f1,
            "gap":            train_f1 - cv_f1,
            "flag_overfit":   (train_f1 - cv_f1) > OVERFIT_GAP,
        })

    overfit_df = pd.DataFrame(overfit_rows).sort_values("gap", ascending=False).reset_index(drop=True)
    print(f"\n[OVERFIT] {ds_id} Train vs CV macro F1 gap (gap > {OVERFIT_GAP} = flagged):")
    display(overfit_df)

    fig, ax = plt.subplots(figsize=(8, 4.5))
    x = np.arange(len(overfit_df))
    ax.bar(x - 0.18, overfit_df["train_macro_f1"], width=0.35,
           label="train macro F1", color="#22c55e", edgecolor="#14532d")
    ax.bar(x + 0.18, overfit_df["cv_macro_f1"],    width=0.35,
           label="CV macro F1",    color="#0ea5e9", edgecolor="#0c4a6e")
    ax.set_xticks(x)
    ax.set_xticklabels(overfit_df["model"], rotation=20, ha="right")
    ax.set_ylim(0, 1.10)
    ax.set_ylabel("Macro F1")
    ax.set_title(f"{ds_id} — Overfitting Check: Train vs. CV Macro F1", fontsize=13, fontweight="bold")
    ax.legend()
    ax.grid(axis="y", alpha=0.35)
    fig.tight_layout()
    fig_path = FIG_DIR / f"overfitting_check_{ds_id}.png"
    fig.savefig(fig_path, dpi=150)
    plt.show()
    print(f"[FIG] Saved -> {fig_path.resolve()}")

    overfit_df.to_csv(ARTIFACT_DIR / f"overfitting_check_{ds_id}.csv", index=False)
    n_flagged = int(overfit_df["flag_overfit"].sum())
    print(f"[OVERFIT] {ds_id}: {n_flagged} model(s) flagged with gap > {OVERFIT_GAP}.")
````

### 7.4 Hyperparameter Study

We sweep **5+ values** for the most influential
hyperparameters and plot the CV macro-F1 curve. Tuning is performed on
the **training partition only**; the locked test set is untouched.

We sweep:

- **MultinomialNB & ComplementNB** — `alpha` smoothing ∈ {0.01, 0.1, 0.5, 1.0, 2.0}
- **LogisticRegression**           — `C` regularisation ∈ {0.25, 0.5, 1.0, 2.0, 4.0}
- **LinearSVC**                    — `C` regularisation ∈ {0.25, 0.5, 1.0, 2.0, 4.0}

The chosen default from (`alpha=1.0`, `C=1.0`) is overlaid on each curve.

This is run **independently for each dataset**.

````python
# ============================================================
# STAGE 7.4: HYPERPARAMETER STUDY (TRAINING-ONLY CV) - PER DATASET
# ============================================================
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB, ComplementNB

sweep_specs = [
    ("MultinomialNB alpha",       "alpha",       [0.01, 0.1, 0.5, 1.0, 2.0],
     lambda v: MultinomialNB(alpha=v)),
    ("ComplementNB alpha",        "alpha",       [0.01, 0.1, 0.5, 1.0, 2.0],
     lambda v: ComplementNB(alpha=v)),
    ("LogisticRegression C",      "C",           [0.25, 0.5, 1.0, 2.0, 4.0],
     lambda v: LogisticRegression(max_iter=2500, class_weight="balanced", C=v,
                                  random_state=RANDOM_STATE)),
    ("LinearSVC C",               "C",           [0.25, 0.5, 1.0, 2.0, 4.0],
     lambda v: LinearSVC(class_weight="balanced", C=v, random_state=RANDOM_STATE)),
]

def cv_macro_f1(estimator, X, y):
    pipe = Pipeline([("tfidf", TfidfVectorizer(**TFIDF_CONFIG)), ("classifier", estimator)])
    scores = cross_validate(pipe, X, y, cv=cv,
                            scoring={"macro_f1": "f1_macro"},
                            n_jobs=N_JOBS)["test_macro_f1"]
    return float(scores.mean()), float(scores.std())

for ds_id in splits.keys():
    train_df = splits[ds_id]["train"]
    X_train = train_df["text"]
    y_train = train_df["label"]
    
    print(f"\n[SWEEP] Running hyperparameter study on {ds_id} ...")

    sweep_rows = []
    for label, hp_name, values, factory in sweep_specs:
        for v in values:
            m, s = cv_macro_f1(factory(v), X_train, y_train)
            sweep_rows.append({"dataset_id": ds_id, "hyperparameter": label, "value": v,
                              "cv_macro_f1_mean": m, "cv_macro_f1_sd": s})

    sweep_df = pd.DataFrame(sweep_rows)
    print(f"[SWEEP] {ds_id} Hyperparameter sensitivity (training-only CV macro F1):")
    display(sweep_df)

    # Plot sweep curves
    fig, axes = plt.subplots(1, 4, figsize=(18, 4))
    for ax, (label, hp_name, values, _factory) in zip(axes, sweep_specs):
        sub = sweep_df[sweep_df["hyperparameter"] == label]
        ax.errorbar(sub["value"], sub["cv_macro_f1_mean"],
                    yerr=sub["cv_macro_f1_sd"], marker="o", capsize=3,
                    color="#0ea5e9", ecolor="#94a3b8", linewidth=2)
        # mark the default from (alpha=1.0 or C=1.0)
        default = 1.0
        if default in sub["value"].values:
            default_row = sub[sub["value"] == default].iloc[0]
            ax.scatter([default], [default_row["cv_macro_f1_mean"]],
                       color="#dc2626", s=120, zorder=5, label="default (alpha=1.0 or C=1.0)")
            ax.legend(loc="lower right")
        ax.set_title(f"{ds_id}: {label}", fontsize=11, fontweight="bold")
        ax.set_xlabel(hp_name)
        ax.set_ylabel("CV macro F1")
        ax.grid(True, alpha=0.35)
        if hp_name == "alpha":
            ax.set_xscale("log")

    fig.suptitle(f"{ds_id} — Hyperparameter Sensitivity (CV macro F1, mean ± SD)",
                 fontsize=14, fontweight="bold")
    fig.tight_layout(rect=[0, 0, 1, 0.95])
    fig_path = FIG_DIR / f"hyperparameter_sweep_{ds_id}.png"
    fig.savefig(fig_path, dpi=150)
    plt.show()
    print(f"[FIG] Saved -> {fig_path.resolve()}")

    sweep_df.to_csv(ARTIFACT_DIR / f"hyperparameter_sweep_{ds_id}.csv", index=False)

    # Choose the best value per hyperparameter and report it.
    best_per_hp = (sweep_df.sort_values("cv_macro_f1_mean", ascending=False)
                              .groupby("hyperparameter", as_index=False)
                              .first())
    print(f"[SWEEP] {ds_id} Best value per hyperparameter:")
    display(best_per_hp)
    best_per_hp.to_csv(ARTIFACT_DIR / f"hyperparameter_best_{ds_id}.csv", index=False)
````

### 7.5 Input Ablation Study (Subject vs Body vs Subject+Body)

We re-run the 5-fold CV benchmark under three input configurations:

1. **`subject` only**  — subject header only.
2. **`body` only**     — the bare email text.
3. **`subject + body`** — the default (current `text` column).

This isolates the contribution of the subject field vs body for spam detection.

````python
# ============================================================
# STAGE 7.5: INPUT ABLATION STUDY (PER DATASET)
# ============================================================
ablation_rows = []

for ds_id in splits.keys():
    train_df = splits[ds_id]["train"].reset_index(drop=True).copy()
    y_train = train_df["label"]
    
    abl_X = {
        "subject_only": train_df["subject"].fillna(""),
        "body_only":    train_df["body"].fillna(""),
        "subject_body": train_df["text"],  # already 'subject: ... \nbody: ...'
    }

    for variant, X in abl_X.items():
        for model_name, pipeline in MODELS.items():
            scores = cross_validate(
                clone(pipeline), X, y_train, cv=cv,
                scoring={"macro_f1": "f1_macro", "accuracy": "accuracy"},
                n_jobs=N_JOBS, error_score="raise")
            ablation_rows.append({
                "dataset_id": ds_id,
                "variant": variant,
                "model":   model_name,
                "accuracy_mean": float(scores["test_accuracy"].mean()),
                "macro_f1_mean": float(scores["test_macro_f1"].mean()),
                "macro_f1_sd":   float(scores["test_macro_f1"].std()),
            })

ablation_df = pd.DataFrame(ablation_rows)
print("[ABL] Input ablation results (training-only CV):")
display(ablation_df)

ablation_df.to_csv(ARTIFACT_DIR / "input_ablation_d2_d3.csv", index=False)

# Heatmap of macro F1 (variant x model) per dataset
for ds_id in splits.keys():
    ds_abl = ablation_df[ablation_df["dataset_id"] == ds_id]
    pivot = ds_abl.pivot(index="model", columns="variant", values="macro_f1_mean")
    vorder = [v for v in ["subject_only", "body_only", "subject_body"] if v in pivot.columns]
    pivot = pivot[vorder]

    fig, ax = plt.subplots(figsize=(8, 4.5))
    sns.heatmap(pivot, annot=True, fmt=".3f", cmap="YlGnBu", cbar_kws={"label": "CV macro F1"},
                vmin=0, vmax=1.0, ax=ax)
    ax.set_title(f"{ds_id} — Input Ablation (CV macro F1)", fontsize=13, fontweight="bold")
    ax.set_xlabel("Input variant")
    ax.set_ylabel("Classifier")
    fig.tight_layout()
    fig_path = FIG_DIR / f"input_ablation_heatmap_{ds_id}.png"
    fig.savefig(fig_path, dpi=150)
    plt.show()
    print(f"[FIG] Saved -> {fig_path.resolve()}")
````

## Stage 8: Cross-Dataset Spam Transfer Evaluation

Direct transfer is valid only for 
harmonized binary spam labels. We train on one spam corpus and test on 
the other **without retraining** to measure domain shift.

- D2 (Enron-Spam) → D3 (SpamAssassin)
- D3 (SpamAssassin) → D2 (Enron-Spam)

We use the best model selected by CV on the **source** dataset.

````python
# ============================================================
# STAGE 8: CROSS-DATASET SPAM TRANSFER EVALUATION
# ============================================================

def cross_dataset_spam(train_id, test_id, model_name):
    """
    Train on full train split of train_id, evaluate on full test split of test_id.
    Returns metrics dict.
    """
    train_df = datasets[train_id]  # full dataset (no split leakage here - this IS the source corpus)
    test_df = datasets[test_id]
    
    expected = {"legitimate", "spam"}
    if set(train_df["label"].unique()) != expected:
        raise ValueError(f"Unexpected labels in {train_id}: {train_df['label'].unique()}")
    if set(test_df["label"].unique()) != expected:
        raise ValueError(f"Unexpected labels in {test_id}: {test_df['label'].unique()}")

    model = clone(MODELS[model_name])
    model.fit(train_df["text"], train_df["label"])
    pred = model.predict(test_df["text"])

    return {
        "train_dataset": train_id,
        "test_dataset": test_id,
        "model": model_name,
        "accuracy": accuracy_score(test_df["label"], pred),
        "macro_f1": f1_score(test_df["label"], pred, average="macro"),
        "weighted_f1": f1_score(test_df["label"], pred, average="weighted"),
    }

# Get best model per dataset from CV results
best_models = {}
for ds_id in splits.keys():
    ranked = cv_results[cv_results["dataset_id"] == ds_id]
    best_models[ds_id] = ranked.iloc[0]["model"]
    print(f"[BEST] {ds_id} best model (by CV macro F1): {best_models[ds_id]}")

cross_rows = [
    cross_dataset_spam("enron_spam", "spamassassin", best_models["enron_spam"]),
    cross_dataset_spam("spamassassin", "enron_spam", best_models["spamassassin"]),
]

cross_df = pd.DataFrame(cross_rows)
print("\n[CROSS-DATASET] Transfer Results:")
display(cross_df)

cross_df.to_csv(ARTIFACT_DIR / "cross_dataset_spam_transfer.csv", index=False)
print(f"[OK] Cross-dataset results saved.")
````

## Stage 9: Model Selection, Fitting, and Locked Test Evaluation

Based on training-only cross-validation macro F1, we select the best model 
per dataset, fit it on the full training set, and evaluate it **once** on 
the locked test set.

````python
# ============================================================
# STAGE 9: SELECT, FIT, AND EVALUATE BEST MODEL ON LOCKED TEST SET
# ============================================================
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, f1_score

selected_models = {}
test_rows = []

for ds_id in splits.keys():
    ranked = cv_results[cv_results["dataset_id"] == ds_id]
    best_name = ranked.iloc[0]["model"]
    print(f"\n[SELECT] {ds_id} best model: {best_name} (CV Macro F1 = {ranked.iloc[0]['macro_f1_mean']:.4f})")

    # Fit on full training set
    best_pipeline = clone(MODELS[best_name])
    train_df = splits[ds_id]["train"]
    best_pipeline.fit(train_df["text"], train_df["label"])
    selected_models[ds_id] = best_pipeline

    # Evaluate on locked test set
    test_df = splits[ds_id]["test"]
    X_test = test_df["text"]
    y_test = test_df["label"]
    y_pred = best_pipeline.predict(X_test)

    test_acc = accuracy_score(y_test, y_pred)
    test_macro_f1 = f1_score(y_test, y_pred, average="macro")
    test_weighted_f1 = f1_score(y_test, y_pred, average="weighted")

    print(f"\n[TEST] {ds_id} Locked Test Set Evaluation for '{best_name}':")
    print(f"  Accuracy    : {test_acc:.4f}")
    print(f"  Macro F1    : {test_macro_f1:.4f}")
    print(f"  Weighted F1 : {test_weighted_f1:.4f}")

    print(f"\nClassification Report ({ds_id}):")
    print(classification_report(y_test, y_pred, digits=4))

    test_rows.append({
        "dataset_id": ds_id,
        "model": best_name,
        "accuracy": test_acc,
        "macro_f1": test_macro_f1,
        "weighted_f1": test_weighted_f1,
    })

    # Persist model
    model_path = MODEL_DIR / f"{ds_id}_{best_name}.joblib"
    joblib.dump(best_pipeline, model_path)
    print(f"[OK] Fitted pipeline saved -> {model_path.resolve()}")

test_results = pd.DataFrame(test_rows)
display(test_results)
test_results.to_csv(ARTIFACT_DIR / "test_results_d2_d3.csv", index=False)
````

### 9.2 Per-Class Results & Confusion Matrices

Per the lab manual §16.5, we emit per-class metrics (support, precision, recall, F1, 
most common confusion) and confusion matrices (counts and row-normalized form) 
for each dataset.

````python
# ============================================================
# STAGE 9.2: PER-CLASS RESULTS TABLE & CONFUSION MATRICES
# ============================================================
from sklearn.metrics import precision_recall_fscore_support

for ds_id in splits.keys():
    best_name = test_results[test_results["dataset_id"] == ds_id].iloc[0]["model"]
    best_pipeline = selected_models[ds_id]
    test_df = splits[ds_id]["test"]
    X_test = test_df["text"]
    y_test = test_df["label"]
    y_pred = best_pipeline.predict(X_test)
    
    labels_sorted = sorted(y_test.unique())
    cm = confusion_matrix(y_test, y_pred, labels=labels_sorted)
    cm_norm = cm.astype("float") / cm.sum(axis=1, keepdims=True)
    np.fill_diagonal(cm_norm, np.nan)  # optional: don't show diagonal in normalized
    
    # Per-class metrics
    prec, rec, f1, support = precision_recall_fscore_support(
        y_test, y_pred, labels=labels_sorted, zero_division=0)

    rows = []
    for i, cls in enumerate(labels_sorted):
        # most-confused-with: argmax of the off-diagonal cell
        row = cm[i].astype(int).copy()
        diag_mask = np.ones(len(labels_sorted), dtype=bool)
        diag_mask[i] = False
        off = row[diag_mask]
        off_labels = [labels_sorted[j] for j, keep in enumerate(diag_mask) if keep]
        most_confused = off_labels[int(np.argmax(off))] if off.size > 0 and off.max() > 0 else "-"
        rows.append({
            "class": cls,
            "support": int(support[i]),
            "precision": float(prec[i]),
            "recall":    float(rec[i]),
            "f1":        float(f1[i]),
            "most_confused_with": most_confused,
        })
    per_class_df = pd.DataFrame(rows)
    print(f"[PER-CLASS] {ds_id} per-class results (locked test):")
    display(per_class_df)
    per_class_df.to_csv(ARTIFACT_DIR / f"per_class_results_{ds_id}.csv", index=False)

    # Confusion Matrix - Counts
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    
    # Counts
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False,
                xticklabels=labels_sorted, yticklabels=labels_sorted, 
                ax=axes[0], annot_kws={"size": 12, "weight": "bold"})
    axes[0].set_title(f"{ds_id} — Test Confusion Matrix (Counts) — {best_name}", 
                      fontsize=13, fontweight="bold")
    axes[0].set_xlabel("Predicted", fontsize=12)
    axes[0].set_ylabel("True", fontsize=12)
    
    # Row-normalized
    sns.heatmap(cm_norm, annot=True, fmt=".2f", cmap="Oranges", cbar=True,
                xticklabels=labels_sorted, yticklabels=labels_sorted,
                ax=axes[1], annot_kws={"size": 11, "weight": "bold"},
                mask=np.diag(cm.sum(axis=1)) > 0)
    axes[1].set_title(f"{ds_id} — Test Confusion Matrix (Row-Normalized, off-diag)", 
                      fontsize=13, fontweight="bold")
    axes[1].set_xlabel("Predicted", fontsize=12)
    axes[1].set_ylabel("True", fontsize=12)
    
    fig.tight_layout()
    fig_path = FIG_DIR / f"test_confusion_matrix_{ds_id}.png"
    fig.savefig(fig_path, dpi=150)
    plt.show()
    print(f"[FIG] Saved -> {fig_path.resolve()}")
````

### 9.3 Bootstrap 95% CI for Locked-Test Macro F1

Per the lab manual §13.1 ("Bootstrap 95% interval — sampling uncertainty 
for locked-test macro F1; report resampling method and count") and §15 
("Uncertainty and abstention"), we resample the **locked test set** with 
replacement (stratified by true label) to quantify sampling uncertainty 
around the single point estimate.

We run **2 000** stratified bootstrap iterations and report the mean
and the 2.5%/97.5% percentile interval for macro F1.

````python
# ============================================================
# STAGE 9.3: BOOTSTRAP 95% CI FOR LOCKED-TEST MACRO F1
# ============================================================
from sklearn.metrics import f1_score as _f1

def stratified_bootstrap_macro_f1(y_true, y_pred, n_boot=2000, seed=RANDOM_STATE):
    """Stratified bootstrap of macro F1 (manual §13.1, Extension Code B1)."""
    rng = np.random.default_rng(seed)
    y_true = np.asarray(y_true); y_pred = np.asarray(y_pred)
    groups = {lab: np.where(y_true == lab)[0] for lab in np.unique(y_true)}
    vals = []
    for _ in range(n_boot):
        idx = np.concatenate([rng.choice(g, size=len(g), replace=True) for g in groups.values()])
        vals.append(_f1(y_true[idx], y_pred[idx], average="macro"))
    arr = np.asarray(vals)
    return float(arr.mean()), float(np.percentile(arr, 2.5)), float(np.percentile(arr, 97.5))

for ds_id in splits.keys():
    best_name = test_results[test_results["dataset_id"] == ds_id].iloc[0]["model"]
    best_pipeline = selected_models[ds_id]
    test_df = splits[ds_id]["test"]
    X_test = test_df["text"]
    y_test = test_df["label"]
    y_pred = best_pipeline.predict(X_test)
    
    test_macro_f1 = f1_score(y_test, y_pred, average="macro")
    boot_mean, boot_lo, boot_hi = stratified_bootstrap_macro_f1(y_test.values, y_pred)
    
    print(f"\n[BOOT] {ds_id} ({best_name})")
    print(f"  Test Macro F1    : {test_macro_f1:.4f}")
    print(f"  Bootstrap Mean   : {boot_mean:.4f}")
    print(f"  95% CI           : [{boot_lo:.4f}, {boot_hi:.4f}]  (n_boot=2000, stratified)")

    # Persist
    with open(ARTIFACT_DIR / f"bootstrap_macro_f1_{ds_id}.json", "w", encoding="utf-8") as fh:
        json.dump({
            "model": best_name,
            "test_macro_f1": test_macro_f1,
            "bootstrap_mean": boot_mean,
            "ci_lower_2p5": boot_lo,
            "ci_upper_97p5": boot_hi,
            "n_boot": 2000,
            "method": "stratified percentile",
            "seed": RANDOM_STATE,
        }, fh, indent=2)
    print(f"[OK] Bootstrap CI saved.")
````
